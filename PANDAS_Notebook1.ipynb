# **Series**

### Basic Series

```python
import numpy as np
import pandas as pd
```

```python
# String 
country = ['India','Pakistan','USA','Nepal','Srilanka']
pd.Series(country)
```

```python
# Integer 
runs = [13,26,39,57,90]
run_ser = pd.Series(runs)
run_ser
```

```python
# Custom Index
marks = [50,67,76,80]
subjects = ['maths','chemistry','physics','Hindi']

pd.Series(marks,index=subjects)
```

```python
# Setting the name of series

pd.Series(marks,index=subjects,name='Aashi ke marks')
```

### Series from dictionary

```python
marks = {
    'maths' : 467,
    'english': 387,
    'science': 297,
    'hindi' : 110
}


marks_series = pd.Series (marks , name = 'HmHogyeKamyab')
marks_series
```

#### sort_values

```python
vk.sort_values() # default ascending order
vk.sort_values(ascending=False) # descending order
vk.sort_values(ascending=False).head(10) # top 10 scores
vk.sort_values(ascending=False).head(1).values[0] # highest score 
# When we apply .values[0] it will give us the value of highest score instead of series
# Method chaining is used for applying multiple methods in a single line of code. It allows us to perform a series of operations on a DataFrame or Series without the need for intermediate variables. This can make the code more concise and easier to read.
# vk.sort_values(inplace=True)
# inplace=True means that the original series will be modified and sorted in place, rather than creating a new sorted series. This can save memory and improve performance when working with large datasets.
```
