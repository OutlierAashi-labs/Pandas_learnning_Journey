## Important DataFrame Functions

### astype 
#### astype() method is used to convert the datatype of a column in a DataFrame. It can be used to convert a column to a different datatype, such as from float to int, or from object to string.

```python
# astype() method is used to convert the datatype of a column in a DataFrame. It can be used to convert a column to a different datatype, such as from float to int, or from object to string.
ipl['ID']= ipl['ID'].astype('int32')


ipl['Season'] = ipl['Season'].astype('category')
ipl['Team1'] = ipl['Team1'].astype('category')
ipl['Team2'] = ipl['Team2'].astype('category')

ipl.info()
```

### value_counts() 
#### value_count method is used to count the number of occurrences of each unique value in a column. It returns a Series containing the counts of unique values.
#### for both (series and dataframe)


```python
movies.value_counts()
students.value_counts()


```

```python
# find which player has won most player of the match -> in final and qualifiers
ipl[~ipl['MatchNumber'].str.isdigit()]['Player_of_Match'].value_counts()     # ~ is not operator, it is used to negate the boolean values. It will return True for all the rows where the MatchNumber is not a digit (i.e., Final and Qualifier matches) and False for all the rows where the MatchNumber is a digit (i.e., group stage matches).






```

```python
# Toss decision plot
ipl.head()
ipl['TossDecision'].value_counts().plot(kind='pie', title='Toss Decision')
```

```python
# how many matches each team has played

ipl['Team1'].value_counts()   
ipl['Team2'].value_counts()

(ipl['Team1'].value_counts() + ipl['Team2'].value_counts()).sort_values(ascending=False)

```
